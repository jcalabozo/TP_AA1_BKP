## 6. Preparación de los datos

### 6.1 División en entrenamiento y prueba

Dividimos **80 % / 20 %** antes de imputar y escalar, para que el conjunto de prueba no influya en
las medianas ni en el escalado (fuga de información).

No apartamos un conjunto de validación fijo: los hiperparámetros se eligen con **validación cruzada
de 5 particiones** sobre el conjunto de entrenamiento (sección 10.2), que aprovecha todos los datos
y da resultados más estables con un dataset de este tamaño.

```python
# Separamos características (X) y variable objetivo (y)
X = datos.drop(columns='MEDV')
y = datos['MEDV']

# División 80 % entrenamiento / 20 % prueba
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=SEMILLA)

print(f'Entrenamiento: {len(X_train)} filas')
print(f'Prueba:        {len(X_test)} filas')
print(f'Media de MEDV — entrenamiento: {y_train.mean():.2f} | '
      f'prueba: {y_test.mean():.2f}')
```

### 6.2 Pipeline de preprocesamiento

Un `Pipeline` encadena la imputación con la mediana y el escalado con `StandardScaler`. Escalamos
porque tanto el descenso por gradiente como la regularización dependen de la escala de las
variables. El `Pipeline` permite además que, en la validación cruzada, la imputación y el escalado
se ajusten dentro de cada partición.

```python
# El pipeline encadena imputación y escalado en un solo objeto
preprocesamiento = Pipeline([
    ('imputador', SimpleImputer(strategy='median')),
    ('escalador', StandardScaler()),
])

# Versión transformada de los datos: la necesitamos porque las funciones de
# descenso por gradiente de la cátedra reciben matrices numéricas, no
# pipelines
X_train_prep = pd.DataFrame(preprocesamiento.fit_transform(X_train),
                            columns=X.columns)
X_test_prep = pd.DataFrame(preprocesamiento.transform(X_test),
                           columns=X.columns)

print('Faltantes tras la imputación:', X_train_prep.isna().sum().sum())
print('Entrenamiento escalado — '
      f'media ≈ {X_train_prep.mean().abs().max():.1e}, '
      f'desvío ≈ {X_train_prep.std().mean():.3f}')
```

### 6.3 Métricas elegidas

- **MSE:** es la función de costo de la regresión lineal y la que registran las funciones de
  descenso por gradiente de la cátedra (Figuras 8 a 10). Está en miles de dólares al cuadrado.
- **RMSE:** es la raíz del MSE, así que está en miles de dólares y penaliza los errores grandes. Es
  la que usamos para elegir hiperparámetros y comparar modelos.
- **MAE:** también en miles de dólares, menos sensible a los errores extremos.
- **R²:** permite comparar con el modelo trivial que predice siempre la media (R² = 0).

**¿Por qué en entrenamiento y en prueba?** Porque juntas permiten distinguir el subajuste (error
alto en los dos conjuntos) del sobreajuste (error bajo en entrenamiento y alto en prueba).

## 7. Regresión lineal múltiple con `LinearRegression`

Es el modelo de referencia.

```python
# Diccionario donde vamos guardando las predicciones de cada modelo, para
# compararlos todos juntos en la sección 11
predicciones = {}

# Entrenamos sobre los datos ya preprocesados
regresion_lineal = LinearRegression().fit(X_train_prep, y_train)

pred_train = regresion_lineal.predict(X_train_prep)
pred_test = regresion_lineal.predict(X_test_prep)
predicciones['LinearRegression'] = (pred_train, pred_test)

print(f'Train   MSE = {mean_squared_error(y_train, pred_train):6.2f}   '
      f'RMSE = {root_mean_squared_error(y_train, pred_train):5.2f}   '
      f'MAE = {mean_absolute_error(y_train, pred_train):5.2f}   '
      f'R² = {r2_score(y_train, pred_train):.3f}')
print(f'Test    MSE = {mean_squared_error(y_test, pred_test):6.2f}   '
      f'RMSE = {root_mean_squared_error(y_test, pred_test):5.2f}   '
      f'MAE = {mean_absolute_error(y_test, pred_test):5.2f}   '
      f'R² = {r2_score(y_test, pred_test):.3f}')
```

```python
# Coeficientes del modelo. Al estar las variables escaladas se pueden
# comparar entre sí: cada uno indica cuánto cambia MEDV si la variable sube
# un desvío estándar
coeficientes = pd.Series(regresion_lineal.coef_, index=X.columns).sort_values()

# Rojo para los coeficientes negativos (bajan el precio) y azul para los
# positivos
colores = ['crimson' if valor < 0 else 'steelblue' for valor in coeficientes]

plt.figure(figsize=(8, 5))
plt.barh(coeficientes.index, coeficientes.values, color=colores)
plt.axvline(0, color='gray', linewidth=1)
plt.xlabel('Coeficiente '
           '(cambio en MEDV por un desvío estándar de la variable)')
plt.title('Figura 6: Coeficientes de LinearRegression')
plt.tight_layout()
plt.show()

print(f'Término independiente: {regresion_lineal.intercept_:.2f} '
      'miles de dólares')
```

```python
# Diagnóstico visual: un modelo perfecto pondría todos los puntos sobre la
# recta y = x
plt.figure(figsize=(7, 6))
plt.scatter(y_train, pred_train, s=18, alpha=0.5, label='Train')
plt.scatter(y_test, pred_test, s=18, alpha=0.8, color='darkorange',
            label='Test')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--',
         label='Predicción perfecta')
plt.xlabel('MEDV real (miles de dólares)')
plt.ylabel('MEDV predicho')
plt.title('Figura 7: Valores reales vs. predichos (LinearRegression)')
plt.legend()
plt.tight_layout()
plt.show()
```

El modelo explica el **66 %** de la varianza en entrenamiento y el **30 %** en prueba (RMSE de 5.55
y 7.61 k\$). La caída podría indicar sobreajuste; lo revisamos con validación cruzada en la sección
11.

En la **Figura 6**, `RM` es la variable que más sube el precio y `DIS`, `LSTAT` y `NOX` las que más
lo bajan. El coeficiente de `RAD` es **positivo** aunque su correlación con el precio es negativa
(−0.34, Figura 4): es un efecto de la multicolinealidad con `TAX`, y muestra que con variables
correlacionadas los coeficientes no deben interpretarse por separado.

En la **Figura 7**, el modelo acierta razonablemente en el rango central, pero falla en las filas
con `MEDV = 50` (la línea vertical de puntos a la derecha) y en algunos casos dispersos.

## 8. Descenso por gradiente

Probamos las tres variantes vistas en clase con las funciones de la cátedra:

| Método | Muestras por actualización | Actualizaciones por época |
|---|---|---|
| Gradient Descent (GD) | todas (428) | 1 |
| Stochastic GD (SGD) | 1 | 428 |
| Mini-Batch GD | 32 | 14 |

Cada método usa la tasa de aprendizaje con la que mejor converge (0.1 en GD, 0.001 en SGD y 0.01 en
Mini-Batch); lo justificamos en la sección 10.1.

Las funciones piden un segundo conjunto para graficar su curva de error; les pasamos el de prueba
solo como referencia visual. Los hiperparámetros se eligen con la curva de entrenamiento (sección
10.1).

```python
# Entrenamos los tres métodos con la misma cantidad de épocas y la misma
# semilla. Cada función grafica sus curvas de error (Figuras 8a a 8c)
EPOCAS = 200
ETIQUETA_PRUEBA = 'Error de prueba'

W_gd, error_train_gd, error_test_gd = gradient_descent(
    X_train_prep, y_train, X_test_prep, y_test,
    lr=0.1, epochs=EPOCAS, semilla=SEMILLA, etiqueta_val=ETIQUETA_PRUEBA,
    titulo='Figura 8a: Error de entrenamiento y prueba vs épocas (GD)')

W_sgd, error_train_sgd, error_test_sgd = stochastic_gradient_descent(
    X_train_prep, y_train, X_test_prep, y_test,
    lr=0.001, epochs=EPOCAS, semilla=SEMILLA, etiqueta_val=ETIQUETA_PRUEBA,
    titulo='Figura 8b: Error de entrenamiento y prueba vs épocas (SGD)')

W_mb, error_train_mb, error_test_mb = mini_batch_gradient_descent(
    X_train_prep, y_train, X_test_prep, y_test,
    lr=0.01, epochs=EPOCAS, batch_size=32, semilla=SEMILLA,
    etiqueta_val=ETIQUETA_PRUEBA,
    titulo=('Figura 8c: Error de entrenamiento y prueba vs épocas '
            '(Mini-Batch GD)'))

# Agrupamos los resultados en tres diccionarios, uno por tipo de resultado,
# para poder recorrerlos cómodamente en los gráficos que siguen
pesos = {'Gradient Descent': W_gd, 'Stochastic GD': W_sgd,
         'Mini-Batch GD': W_mb}
error_train = {'Gradient Descent': error_train_gd,
               'Stochastic GD': error_train_sgd,
               'Mini-Batch GD': error_train_mb}
error_test = {'Gradient Descent': error_test_gd,
              'Stochastic GD': error_test_sgd,
              'Mini-Batch GD': error_test_mb}

for nombre, W in pesos.items():
    predicciones[nombre] = (predecir(X_train_prep, W),
                            predecir(X_test_prep, W))
```

```python
# Los tres métodos juntos, contra el MSE mínimo alcanzable (el de la
# solución analítica). Usamos escala logarítmica porque el error inicial,
# con pesos aleatorios, es mucho mayor que el final
epocas = range(1, EPOCAS + 1)
mse_minimo = mean_squared_error(y_train,
                                regresion_lineal.predict(X_train_prep))

plt.figure(figsize=(10, 5.5))
for nombre in pesos:
    plt.plot(epocas, error_train[nombre], label=nombre)
plt.axhline(mse_minimo, color='black', linestyle='--',
            label=f'LinearRegression (mínimo = {mse_minimo:.2f})')
plt.yscale('log')
plt.xlabel('Época')
plt.ylabel('MSE de entrenamiento (escala logarítmica)')
plt.title('Figura 9: Velocidad de convergencia de los tres métodos')
plt.legend()
plt.tight_layout()
plt.show()

# Buscamos en qué época cada método baja del 1 % por encima del mínimo
# analítico
umbral = mse_minimo * 1.01

for nombre in pesos:
    epoca = None
    for numero_epoca, error in enumerate(error_train[nombre], start=1):
        if error <= umbral:
            epoca = numero_epoca
            break
    print(f'{nombre:18s} MSE final = {error_train[nombre][-1]:7.3f}   '
          f'llega al 1 % del mínimo en la época {epoca}')
```

```python
# Coeficientes de cada método frente a la solución analítica.
# Las funciones de la cátedra devuelven el término independiente en la
# primera posición de W, así que armamos la columna de LinearRegression con
# el mismo orden
pesos_analiticos = np.concatenate([[regresion_lineal.intercept_],
                                   regresion_lineal.coef_])

comparacion_pesos = pd.DataFrame({'LinearRegression': pesos_analiticos},
                                 index=['intercepto'] + list(X.columns))
for nombre, W in pesos.items():
    comparacion_pesos[nombre] = W.ravel()

display(comparacion_pesos.round(3))

# Cuánto se aparta cada método de la solución exacta, coeficiente por
# coeficiente
diferencias = comparacion_pesos.drop(columns='LinearRegression')
for nombre in pesos:
    diferencia = (diferencias[nombre]
                  - comparacion_pesos['LinearRegression']).abs().max()
    print(f'{nombre:18s} diferencia máxima con LinearRegression = '
          f'{diferencia:.4f}')
```

```python
# Métricas de los cuatro modelos ajustados hasta acá
tabla = []
for nombre, (p_train, p_test) in predicciones.items():
    tabla.append({
        'Modelo': nombre,
        'MSE (Train)': mean_squared_error(y_train, p_train),
        'RMSE (Train)': root_mean_squared_error(y_train, p_train),
        'MAE (Train)': mean_absolute_error(y_train, p_train),
        'R² (Train)': r2_score(y_train, p_train),
        'MSE (Test)': mean_squared_error(y_test, p_test),
        'RMSE (Test)': root_mean_squared_error(y_test, p_test),
        'MAE (Test)': mean_absolute_error(y_test, p_test),
        'R² (Test)': r2_score(y_test, p_test),
    })
display(pd.DataFrame(tabla).round(3))
```

**¿Algún cambio respecto de `LinearRegression`?** **En los resultados, prácticamente ninguno:** los
tres métodos llegan a un MSE de entrenamiento casi igual al de `LinearRegression` (30.79) y sus
coeficientes no se apartan más de 0.07, porque minimizan la misma función de costo.

Las diferencias están en el camino (Figuras 8a a 8c y 9):

- **Velocidad:** SGD llega al 1 % del mínimo en 10 épocas, Mini-Batch en 25 y GD en 31, por la
  cantidad de actualizaciones de pesos por época (428, 14 y 1).
- **Forma de las curvas:** la de GD es suave; las de SGD y Mini-Batch oscilan porque cada paso usa
  pocas muestras.
- **Precisión final:** GD llega prácticamente al mínimo (30.7870 frente a 30.7867); SGD y Mini-Batch
  quedan apenas por encima (30.805 y 30.800).
- **Error de prueba:** en las tres curvas llega a su mínimo (cerca de 50) en las primeras 15 épocas
  y después sube hasta estabilizarse cerca de 58, el valor de `LinearRegression`. No usamos ese
  mínimo para elegir la cantidad de épocas, porque sería elegir mirando el conjunto de prueba.

Para un dataset de este tamaño, `LinearRegression` es preferible: da la solución exacta y no tiene
hiperparámetros.

## 9. Modelos con regularización: Ridge, Lasso y Elastic Net

Probamos los tres con el `alpha` por defecto de scikit-learn (`alpha = 1`), dentro de un `Pipeline`
con el preprocesamiento, y medimos su error con validación cruzada. El `alpha` se optimiza en la
sección 10.2.

```python
# Entrenamos los modelos regularizados con el alpha por defecto y medimos
# su error por validación cruzada de 5 particiones. Incluimos
# LinearRegression como referencia (vuelve a entrenarse, pero da
# exactamente lo mismo que en la sección 7).
rmse_cv = {}

for nombre, estimador in [('LinearRegression', LinearRegression()),
                          ('Ridge (α=1)', Ridge(alpha=1.0)),
                          ('Lasso (α=1)', Lasso(alpha=1.0, max_iter=50_000)),
                          ('ElasticNet (α=1)',
                           ElasticNet(alpha=1.0, max_iter=50_000))]:
    flujo = Pipeline([('preprocesamiento', preprocesamiento),
                      ('regresor', estimador)])

    # cross_val_score devuelve el RMSE negativo (scikit-learn siempre
    # maximiza), por eso el signo
    puntajes = cross_val_score(flujo, X_train, y_train, cv=5,
                               scoring='neg_root_mean_squared_error')
    rmse_cv[nombre] = -puntajes.mean()

    flujo.fit(X_train, y_train)
    predicciones[nombre] = (flujo.predict(X_train), flujo.predict(X_test))

    coeficientes_no_nulos = (flujo.named_steps['regresor'].coef_ != 0).sum()
    print(f'{nombre:18s} RMSE de validación cruzada = '
          f'{rmse_cv[nombre]:.3f} ± {puntajes.std():.3f}'
          f'   coeficientes no nulos: {coeficientes_no_nulos} de 13')
```

```python
# Métricas de los modelos regularizados frente al modelo base
tabla = []
for nombre in ['LinearRegression', 'Ridge (α=1)', 'Lasso (α=1)',
               'ElasticNet (α=1)']:
    p_train, p_test = predicciones[nombre]
    tabla.append({
        'Modelo': nombre,
        'MSE (Train)': mean_squared_error(y_train, p_train),
        'RMSE (Train)': root_mean_squared_error(y_train, p_train),
        'MAE (Train)': mean_absolute_error(y_train, p_train),
        'R² (Train)': r2_score(y_train, p_train),
        'MSE (Test)': mean_squared_error(y_test, p_test),
        'RMSE (Test)': root_mean_squared_error(y_test, p_test),
        'MAE (Test)': mean_absolute_error(y_test, p_test),
        'R² (Test)': r2_score(y_test, p_test),
    })
display(pd.DataFrame(tabla).round(3))
```

Con `alpha = 1`:

- **Ridge** queda prácticamente igual que `LinearRegression`.
- **Lasso** anula 7 de los 13 coeficientes. Ajusta peor en entrenamiento (R² de 0.66 a 0.56) pero
  mejor en prueba (de 0.30 a 0.49).
- **Elastic Net** anula 2 coeficientes y da un resultado parecido al de Lasso.

Sin embargo, por validación cruzada Lasso y Elastic Net son los **peores** (6.48 y 6.45 frente a
5.83): su mejora en prueba no se repite en las 5 particiones. Por eso elegimos hiperparámetros con
validación cruzada y no con el conjunto de prueba (sección 11).

## 10. Optimización de hiperparámetros

### 10.1 Hiperparámetros del descenso por gradiente

La tasa de aprendizaje y el tamaño de lote determinan si el algoritmo llega al mínimo, así que los
elegimos mirando la curva de entrenamiento.

```python
# Barrido de tasas de aprendizaje (panel izquierdo) y de tamaños de lote
# (panel derecho)
# Con una tasa muy grande el error se dispara: recortamos la curva en TOPE
# para que entre en el gráfico
TOPE = 1e6
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# np.errstate silencia los avisos de desborde numérico que produce la tasa
# que diverge
with np.errstate(over='ignore', invalid='ignore'):
    for tasa in [0.0001, 0.001, 0.01, 0.05, 0.1, 0.3]:
        W, errores, _ = gradient_descent(
            X_train_prep, y_train, X_test_prep, y_test,
            lr=tasa, epochs=EPOCAS, semilla=SEMILLA, graficar=False)
        # Reemplazamos los infinitos y recortamos los valores enormes en TOPE
        curva = np.nan_to_num(errores, nan=TOPE, posinf=TOPE).clip(max=TOPE)
        axes[0].plot(epocas, curva, label=f'lr = {tasa}')

for tamano in [1, 8, 32, 128, len(X_train_prep)]:
    W, errores, _ = mini_batch_gradient_descent(
        X_train_prep, y_train, X_test_prep, y_test,
        lr=0.01, epochs=EPOCAS, batch_size=tamano,
        semilla=SEMILLA, graficar=False)
    axes[1].plot(epocas, errores, label=f'batch_size = {tamano}')

for ax, titulo in zip(axes, ['Tasa de aprendizaje (Gradient Descent)',
                             'Tamaño de lote (Mini-Batch, lr = 0.01)']):
    ax.axhline(mse_minimo, color='black', linestyle='--',
               label='Mínimo alcanzable')
    ax.set_yscale('log')
    ax.set_xlabel('Época')
    ax.set_title(titulo)
    ax.legend(fontsize=9)

axes[0].set_ylabel('MSE de entrenamiento (escala logarítmica)')
fig.suptitle('Figura 10: Efecto de los hiperparámetros del descenso por '
             'gradiente', fontsize=14)
fig.tight_layout()
plt.show()
```

```python
# MSE de entrenamiento después de 200 épocas, con la misma grilla de tasas
# para los tres métodos
metodos = {
    'Gradient Descent': (gradient_descent, {}),
    'Stochastic GD': (stochastic_gradient_descent, {}),
    'Mini-Batch GD': (mini_batch_gradient_descent, {'batch_size': 32}),
}
mse_final = {}
with np.errstate(over='ignore', invalid='ignore'):
    for nombre, (metodo, extra) in metodos.items():
        for tasa in [0.001, 0.01, 0.1]:
            _, errores, _ = metodo(X_train_prep, y_train, X_test_prep, y_test,
                                   lr=tasa, epochs=EPOCAS, semilla=SEMILLA,
                                   graficar=False, **extra)
            mse_final.setdefault(f'lr = {tasa}', {})[nombre] = errores[-1]

# Cuando el método diverge, el error termina en infinito o NaN
tabla_tasas = pd.DataFrame(mse_final)
display(tabla_tasas.round(2).astype(object)
        .where(np.isfinite(tabla_tasas), 'diverge'))
```

**Tasa de aprendizaje (panel izquierdo):**

- `lr = 0.0001` y `0.001`: demasiado chicas. A las 200 épocas el MSE sigue en varios cientos, lejos
  del mínimo de 30.79.
- `lr = 0.01`: baja mucho más rápido, pero todavía más despacio que las siguientes.
- `lr = 0.05` y `0.1`: convergen al mínimo en unas decenas de épocas. Elegimos `lr = 0.1` para GD:
  de los valores probados, es el más grande que converge.
- `lr = 0.3`: el algoritmo **diverge** y el error se dispara.

**Una tasa para cada método (tabla anterior):** ninguna tasa sirve para los tres. Con 0.1, GD
converge pero SGD diverge; con 0.01, GD todavía no llega al mínimo y SGD se estanca en 39.77; con
0.001, SGD converge pero GD queda lejísimos. SGD necesita la tasa más chica porque actualiza los
pesos 428 veces por época, con el gradiente de una sola muestra. Por eso usamos 0.1 en GD, 0.001 en
SGD y 0.01 en Mini-Batch.

**Tamaño de lote (panel derecho):** con `batch_size = 1` el error baja rápido al principio pero se
estanca por encima del resto; con `batch_size = 428` (todo el conjunto) la curva es suave pero
lenta. Los valores intermedios combinan velocidad y estabilidad.

### 10.2 Hiperparámetros de Ridge, Lasso y Elastic Net

Buscamos el mejor `alpha` con `GridSearchCV` sobre 25 valores entre 0.001 y 1000, con validación
cruzada de 5 particiones sobre el conjunto de entrenamiento. En Elastic Net buscamos además
`l1_ratio` (0.1, 0.3, 0.5, 0.7 y 0.9).

```python
# 25 valores de alpha entre 0.001 y 1000, espaciados de forma logarítmica
ALPHAS = np.logspace(-3, 3, 25)
busquedas = {}

# En Elastic Net también buscamos l1_ratio, que define cuánto pesa la
# penalización L1 frente a la L2
for nombre, estimador, grilla in [
        ('Ridge', Ridge(), {}),
        ('Lasso', Lasso(max_iter=50_000), {}),
        ('ElasticNet', ElasticNet(max_iter=50_000),
         {'regresor__l1_ratio': [0.1, 0.3, 0.5, 0.7, 0.9]})]:
    flujo = Pipeline([('preprocesamiento', preprocesamiento),
                      ('regresor', estimador)])

    # 'regresor__alpha' significa: el parámetro alpha del paso llamado
    # 'regresor'
    busqueda = GridSearchCV(flujo, {'regresor__alpha': ALPHAS, **grilla},
                            cv=5, scoring='neg_root_mean_squared_error',
                            return_train_score=True).fit(X_train, y_train)
    busquedas[nombre] = busqueda

    # best_estimator_ es el pipeline ya reentrenado con todo X_train y los
    # mejores hiperparámetros
    mejor_alpha = busqueda.best_params_['regresor__alpha']
    mejor_l1 = busqueda.best_params_.get('regresor__l1_ratio')
    etiqueta = f'{nombre} (α={mejor_alpha:.3g})'
    if mejor_l1 is not None:
        etiqueta = f'{nombre} (α={mejor_alpha:.3g}, l1_ratio={mejor_l1})'
    predicciones[etiqueta] = (busqueda.predict(X_train),
                              busqueda.predict(X_test))
    rmse_cv[etiqueta] = -busqueda.best_score_

    regresor = busqueda.best_estimator_.named_steps['regresor']
    coeficientes_no_nulos = (regresor.coef_ != 0).sum()
    print(f'{nombre:12s} mejor alpha = {mejor_alpha:8.4f}   '
          f'RMSE de validación cruzada = {rmse_cv[etiqueta]:.3f}   '
          f'coeficientes no nulos: {coeficientes_no_nulos} de 13')
    if mejor_l1 is not None:
        print(f'{"":12s} mejor l1_ratio = {mejor_l1}')
```

```python
# Curva de error en función de alpha, directamente desde los resultados de
# GridSearchCV. La banda sombreada es ±1 desvío entre las 5 particiones
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5), sharey=True)

for ax, (nombre, busqueda) in zip(axes, busquedas.items()):
    # cv_results_ trae el error de cada alpha; los puntajes vienen negados,
    # por eso el signo
    resultados_cv = pd.DataFrame(busqueda.cv_results_)
    titulo = nombre
    # En Elastic Net graficamos la curva del mejor l1_ratio
    if 'param_regresor__l1_ratio' in resultados_cv:
        mejor_l1 = busqueda.best_params_['regresor__l1_ratio']
        resultados_cv = resultados_cv[
            resultados_cv['param_regresor__l1_ratio'] == mejor_l1]
        titulo = f'{nombre} (l1_ratio = {mejor_l1})'
    alphas = resultados_cv['param_regresor__alpha'].astype(float)
    rmse_train = -resultados_cv['mean_train_score']
    rmse_validacion = -resultados_cv['mean_test_score']
    desvio = resultados_cv['std_test_score']
    mejor_alpha = busqueda.best_params_['regresor__alpha']

    ax.plot(alphas, rmse_train, marker='o', markersize=3, label='Train')
    ax.plot(alphas, rmse_validacion, marker='o', markersize=3, label='CV')
    ax.fill_between(alphas, rmse_validacion - desvio,
                    rmse_validacion + desvio, alpha=0.2,
                    color='darkorange', label='±1 desvío entre particiones')
    ax.axvline(mejor_alpha, color='crimson', linestyle='--',
               label=f'Mejor α = {mejor_alpha:.3g}')
    ax.set_xscale('log')
    ax.set_xlabel('alpha (escala logarítmica)')
    ax.set_title(titulo)
    ax.legend(fontsize=8)

axes[0].set_ylabel('RMSE (miles de dólares)')
axes[0].set_ylim(4, 11)
fig.suptitle('Figura 11: Error en función de la fuerza de regularización',
             fontsize=14)
fig.tight_layout()
plt.show()
```

```python
# Efecto de selección de variables de Lasso: cuántos coeficientes
# sobreviven a cada alpha
for alpha in [0.001, 0.032, 0.316, 1, 3.162, 10]:
    flujo = Pipeline([('preprocesamiento', preprocesamiento),
                      ('regresor', Lasso(alpha=alpha, max_iter=50_000))])
    flujo.fit(X_train, y_train)
    coeficientes_no_nulos = (flujo.named_steps['regresor'].coef_ != 0).sum()
    print(f'Lasso con α = {alpha:<7} '
          f'coeficientes no nulos: {coeficientes_no_nulos:2d} de 13')
```

En la **Figura 11**, el RMSE de entrenamiento crece con `alpha`, y el de validación cruzada se
mantiene casi plano hasta que `alpha` se vuelve grande. El mejor `alpha` de Ridge (≈ 17.8) baja el
RMSE de validación cruzada de 5.829 a 5.813: apenas **0.016 k\$**, frente a un desvío entre
particiones de **±1.1 k\$**. Es decir, **en este dataset la regularización no produce una mejora
real**.

La tabla de Lasso muestra su **selección de variables**: al aumentar `alpha` va anulando
coeficientes hasta quedarse sin ninguno. Con el mejor `alpha`, Lasso conserva 12 de los 13
coeficientes, y Ridge y Elastic Net los 13. En Elastic Net, el mejor `l1_ratio` es 0.1, el más
cercano a Ridge de los que probamos, y su RMSE de validación cruzada (5.814) queda prácticamente
igual al de Ridge (5.813).

## 12. Conclusiones

- **Datos:** eliminamos las 21 filas sin `MEDV` e imputamos los demás faltantes (alrededor de un 4 %
  por columna) con la mediana, por los valores atípicos. Además, `RAD` y `TAX` están muy

- **Preprocesamiento:** dividir antes de imputar y escalar, y hacerlo dentro de un `Pipeline`, evita

- **Métodos de ajuste:** `LinearRegression` y las tres variantes de descenso por gradiente llegan a
  la misma solución (coeficientes a menos de 0.07). Cambia el camino: GD es suave, SGD es rápido
  pero oscila y Mini-Batch queda en el medio (Figuras 9 y 10). Para este tamaño de dataset, la
  solución analítica es preferible.
- **Hiperparámetros:** la tasa de aprendizaje define si el gradiente converge (con 0.3 diverge,
  Figura 10). El `alpha` de la regularización casi no cambia el resultado.
- **Mejor modelo:** Ridge con `alpha ≈ 17.8` por validación cruzada, pero todos los modelos lineales
  son equivalentes (0.016 k\$ de mejora contra 1.1 k\$ de desvío). Lasso con `alpha = 1`, el mejor
  en prueba y el peor en validación cruzada, muestra por qué no hay que elegir mirando el conjunto
  de prueba.
- **Límite:** el sobreajuste es chico; lo que limita al modelo es que la relación no es lineal. Para
  mejorar habría que transformar las variables o agregar términos polinómicos.
